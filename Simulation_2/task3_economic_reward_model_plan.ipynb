# Task 3 Plan: Economic Reward Model

## Objective
Define a reward aligned with business economics: escalation cost, churn risk, and interaction cost, with clear success incentives.

## Plan
1. Write reward decomposition and sign conventions.
2. Define tier-dependent escalation costs.
3. Define churn probability function and expected churn loss.
4. Define terminal and step rewards.
5. Add normalization and clipping for stable learning.

## Reward Decomposition
At step t:

$$
R_t = -C_e(\text{tier}) \cdot \mathbf{1}_{\text{Escalate}} - L_t - \lambda_{turn} + \eta \cdot \mathbf{1}_{\text{Success}}
$$

where\n

$$
L_t = E[\text{ChurnLoss}_t] = P_{churn,t} \cdot V(\text{tier}, \text{customervalueweight})
$$

## Escalation Cost
Tier-dependent cost table (example):
- Free: low or disabled escalation policy
- Pro: C_e = c_pro
- Business: C_e = c_business
- Enterprise: C_e = c_enterprise

Use normalized values for RL and keep a conversion map to real currency for reporting.

## Churn Probability Model
Use a bounded logistic form:

$$
P_{churn,t} = \sigma\left(\gamma_0 + \gamma_f f_t + \gamma_k k_t + \gamma_n n_t + \gamma_v \cdot \text{customervalueweight} + \gamma_u \cdot \mathbf{1}_{\text{unresolved pressure}}\right)
$$

Monotonic assumptions:
- higher frustration -> higher churn probability
- higher failed streak -> higher churn probability
- longer unresolved trajectories -> higher churn probability

## Customer Value at Risk
Define:

$$
V = V_{base}(\text{tier}) \cdot (1 + \kappa \cdot \text{customervalueweight})
$$

This allows tier-level value and within-tier refinement.

## Step vs Terminal Rewards
- Step penalties: turn penalty and optional small compute/inference cost
- Terminal success: positive bonus eta
- Terminal failure: expected churn loss applied strongly
- Escalation terminal: escalation cost plus residual churn adjustment if needed

## Recommended Final Reward Form

$$
R_t = -C_e \cdot \mathbf{1}_{esc} - \lambda_{turn} - \lambda_{inf} + \eta \cdot \mathbf{1}_{success} - \omega \cdot P_{churn,t}V
$$

with optional terminal boost/penalty terms for clearer credit assignment.

## Stability and Calibration
- Normalize reward scale to roughly [-5, +5] initially
- Clip extreme values to control variance
- Run sensitivity sweeps over lambda_turn, eta, omega
- Validate that policy incentives match business intuition

## Deliverables for Task 3
- complete reward equation
- calibrated churn model
- mapping from simulated reward to economic KPIs
