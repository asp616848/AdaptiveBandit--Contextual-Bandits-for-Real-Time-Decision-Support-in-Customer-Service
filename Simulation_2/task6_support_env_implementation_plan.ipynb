## Step Method Contract

Output: observation, reward, done, info.

- reward decomposition
- transition events (success/fail/escalate/close)
- optional text artifacts if hybrid layer enabled

## Reproducibility
- use a dedicated RNG object in environment state
- all stochastic draws come from that RNG
- expose seed in reset and constructor

## Safety and Robustness
- validate actions
- clip bounded variables
- hard-stop at T_max
- avoid silent NaN propagation
