# Task 5 Plan: Hybrid Simulator System Design

## Objective
Design a hybrid simulator where hidden-state transitions are authoritative, while retrieval and optional LLM text are realism layers only.

## Plan
1. Define the canonical step pipeline.
2. Separate state engine from text engine interfaces.
3. Define data contracts for each component.
4. Add safety and fallback behavior.

## Canonical Step Flow
Agent action
-> Hidden-state transition engine (authoritative)
-> Optional retrieval of similar contexts
-> Optional text generation conditioned on updated state
-> Reward computation
-> Return observation, reward, done, info

## Core Principle
Only the state engine can change latent variables. Retrieval and LLM output cannot directly mutate state.

## Component Contracts
### A) State Engine
Input: current latent state and action.
Output: next latent state, event flags, termination status.

### B) Retrieval Layer (optional)
Input: state summary, persona, problem_type, action history.
Output: top-k similar snippets and metadata.

### C) Text Layer (optional)
Input: updated state plus retrieval context.
Output: synthetic customer utterance and confidence metadata.

### D) Reward Engine
Input: transition outcome and economic parameters.
Output: scalar reward and decomposition dictionary.

## Observation Design
Default RL observation should be structured numeric state vector.
Text can be attached in info for analysis, not required for control.

## Fallback and Robustness
- If retrieval fails: continue with state-only simulation.
- If text generation fails: return template text and proceed.
- If generated text contradicts state: state remains source of truth.

## Logging Requirements
Each step should log:
- pre and post state snapshots
- chosen action
- derived p_t and success/failure event
- reward and decomposition
- retrieval ids and generation metadata if used

## Deliverables for Task 5
- final architecture diagram and interfaces
- strict state-authority rules
- fallback and monitoring protocol
