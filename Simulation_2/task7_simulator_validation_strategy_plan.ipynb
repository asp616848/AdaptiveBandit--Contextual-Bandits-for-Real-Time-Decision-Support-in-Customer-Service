# Task 7 Plan: Validation Strategy

## Objective
Define a validation framework to test simulator realism, policy relevance, and economic consistency.

## Plan
1. Define realism metrics against dataset patterns.
2. Define transition sanity tests.
3. Define policy-facing stress tests.
4. Define reporting and acceptance thresholds.

## A) Distributional Realism Checks
Compare simulated vs empirical distributions for:
- turn counts
- resolution rate
- escalation frequency
- frustration trajectories

Recommended statistics:
- KS statistic for continuous features
- chi-square for categorical proportions
- KL divergence for action or outcome distributions

## B) Transition-Logic Sanity Checks
- AskInfo should increase information on average
- failed ProvideSolution should increase frustration and failed_streak
- effective AffectiveRepair should reduce frustration on average
- Escalate and valid Close should terminate

## C) Reward-Economics Sanity Checks
- high-value unresolved trajectories should incur higher expected loss
- unnecessary escalation should be penalized
- successful low-cost automation should receive higher net reward

## D) Policy Stress Tests
Evaluate simple policies to check environment signal quality:
1. Always Escalate
2. Never Escalate
3. Rule threshold on p_t
4. Random policy

Expected ordering should be economically sensible.

## E) Calibration and Ablation
- with vs without persona effects
- with vs without value_weight
- with vs without stochastic transition noise
- sensitivity over key coefficients

## Acceptance Criteria (initial)
- key metric deltas within pre-set tolerance bands
- no pathological trajectories (for example, oscillatory frustration explosions)
- stable training curves for baseline RL runs

## Reporting Package
Produce:
- realism comparison tables
- trajectory plots by persona and tier
- reward decomposition diagnostics
- known limitations and calibration gaps

## Deliverables for Task 7
- validation protocol checklist
- metric definitions and threshold table
- baseline stress-test results template
