# Task 4 Plan: Persona Integration

## Objective
Integrate persona clusters into transition dynamics so behavior is heterogeneous but interpretable.

## Plan
1. Define persona parameter interface from clustering output.
2. Map persona parameters to transition sensitivities.
3. Define default persona archetypes for initialization.
4. Add calibration checks and fairness checks.

## Persona Parameter Interface
For each conversation, assign persona c with soft or hard assignment, then map to:

$$
\psi_c = (rho_c, sigma_c, tau_c)
$$

- rho: patience
- sigma: sensitivity to negative outcomes
- tau: tolerance to repeated failures

## Dynamics Mapping
Use persona in transition equations only through coefficients to keep model interpretable.

Examples:
- higher rho -> slower frustration growth
- higher sigma -> larger failure shock in frustration
- lower tau -> stronger repeated-failure amplification

## Suggested Archetypes (initial, to be replaced by GMM fit)
1. Cooperative: high rho, low sigma, high tau
2. Impatient: low rho, high sigma, medium tau
3. Escalation-prone: medium rho, high sigma, low tau
4. Silent-dropoff-prone: low rho, medium sigma, low tau

## Soft-Assignment Option
If GMM gives posterior weights w_c, use expectation form:

$$
\bar{rho} = \sum_c w_c rho_c, \quad \bar{sigma} = \sum_c w_c sigma_c, \quad \bar{tau} = \sum_c w_c tau_c
$$

This reduces discontinuities across personas.

## Calibration Procedure
1. Fit GMM on persona feature table from pipeline.
2. Check cluster interpretability with summary stats.
3. Map each cluster to parameter priors.
4. Validate downstream effects on escalation, resolution, and turn count distributions.

## Risk Controls
- Keep persona effects bounded to avoid unrealistic trajectories
- Avoid direct protected-attribute proxies
- Report performance by persona group to detect simulator bias

## Deliverables for Task 4
- persona-to-dynamics mapping table
- calibrated persona priors
- validation plots showing heterogeneity realism
