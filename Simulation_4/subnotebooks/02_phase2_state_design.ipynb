```python
from pathlib import Path
import json
import subprocess
import pandas as pd

root = Path.cwd()
if root.name != 'Simulation_4':
    root = root / 'Simulation_4'

script = root / 'scripts' / 'phase2_state_vector_spec.py'
print('Running:', script)
subprocess.run(['python', str(script)], check=True)
```
