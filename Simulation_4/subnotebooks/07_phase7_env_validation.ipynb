# Phase 7 Environment Validation

This notebook validates the Phase 7 `SupportEnv` implementation and runs the smoke test suite.

```python
from pathlib import Path
import sys

repo_root = Path.cwd()
if not (repo_root / 'Simulation_4').exists():
    repo_root = repo_root.parent

sys.path.insert(0, str(repo_root))
repo_root
```

```python
from Simulation_4.env import SupportEnv

artifacts_root = repo_root / 'Simulation_4' / 'artifacts'
env = SupportEnv(artifacts_root=str(artifacts_root), nlg_enabled=False)
obs, info = env.reset(seed=42)
obs, info
```

```python
import numpy as np

rng = np.random.default_rng(42)
trace = []
total_reward = 0.0

for t in range(1, 11):
    action = int(rng.integers(0, 5))
    obs, reward, done, _, info = env.step(action)
    total_reward += reward
    trace.append({
        'turn': t,
        'action': env.ACTION_NAMES[action],
        'reward': reward,
        'done': done,
        'terminal_type': info.get('last_transition_outcome', {}).get('terminal_type')
    })
    if done:
        break

trace, total_reward
```

```python
import subprocess

smoke_script = repo_root / 'Simulation_4' / 'scripts' / 'phase 7' / 'phase7_smoke_test.py'
result = subprocess.run([sys.executable, str(smoke_script)], capture_output=True, text=True)
print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
    raise RuntimeError(f'Smoke test failed with exit code {result.returncode}')
```
