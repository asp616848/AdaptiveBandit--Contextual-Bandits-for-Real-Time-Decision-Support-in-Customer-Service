# Phase 3 Transition Dynamics Calibration
Run the calibration script and inspect all generated outputs under `Simulation_4/artifacts/phase3/`.

```python
from pathlib import Path
import subprocess
import json
import pandas as pd

root = Path.cwd()
if (root / 'Simulation_4').exists():
    REPO_ROOT = root
else:
    REPO_ROOT = root.parent.parent

SCRIPT = REPO_ROOT / 'Simulation_4' / 'scripts' / 'phase 3' / 'phase3_transition_calibration.py'
OUT_DIR = REPO_ROOT / 'Simulation_4' / 'artifacts' / 'phase3'
SCRIPT, OUT_DIR
```

```python
res = subprocess.run(['python', str(SCRIPT)], cwd=REPO_ROOT, text=True, capture_output=True)
print('exit_code:', res.returncode)
print(res.stdout)
if res.stderr:
    print('--- stderr ---')
    print(res.stderr)
```

```python
files = sorted([p.name for p in OUT_DIR.glob('*.json')])
files
```

```python
transition = json.loads((OUT_DIR / 'transition_calibration.json').read_text())
transition
```

```python
dropout_df = pd.DataFrame(transition['dropout']['state_evaluations'])
dropout_df
```

```python
ask_good = pd.DataFrame(transition['frustration_deltas']['askinfo_good'])
ask_bad = pd.DataFrame(transition['frustration_deltas']['askinfo_bad'])
provide_fail = pd.DataFrame(transition['frustration_deltas']['provide_solution_failure'])
repair_expected = pd.DataFrame(transition['frustration_deltas']['affective_repair_expected'])

print('AskInfo good scenario:')
display(ask_good)
print('AskInfo bad scenario:')
display(ask_bad)
print('ProvideSolution failure deltas:')
display(provide_fail[['persona', 'streak_0', 'streak_1', 'streak_2', 'streak_3', 'cumulative_delta_after_3_failures']])
print('AffectiveRepair expected impact:')
display(repair_expected[['persona', 'p_repair', 'expected_delta_f_per_repair', 'repair_actions_to_reduce_f_by_0_20']])
```
