# Phase 8 RAG Validation
Validate document chunking, policy retrieval, and episode context generation for Lumo support scenarios.

```python
from pathlib import Path

from Simulation_4.rag.document_store import DocumentStore
from Simulation_4.rag.lumo_rag import LumoRAG

project_root = Path.cwd()
if not (project_root / 'Simulation_4').exists():
    project_root = project_root.parent

rag_dir = project_root / 'Simulation_4' / 'rag'
store = DocumentStore(str(rag_dir / 'documents'))
rag = LumoRAG(rag_dir=str(rag_dir), index_path=str(rag_dir / 'index'), enabled=True)

print('chunks:', len(store.get_all_chunks()))
```

```python
queries = [
    'duplicate charge in same billing period',
    'lost phone and cannot access 2fa',
    'cancel subscription and export data',
]
for q in queries:
    results = rag.retriever.retrieve(q, top_k=2, doc_type_filter='support_policies')
    print('---')
    print('query:', q)
    for r in results:
        print(' ', r['section_title'], 'score=', round(r['similarity_score'], 4))
```
